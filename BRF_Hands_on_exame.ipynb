```python
def group_by_company(consultants):
    grouped = {}
    for consultant in consultants:
        company = consultant['company']
        if company not in grouped:
            grouped[company] = []
        grouped[company].append(consultant)
    return grouped

consultants = [
    {'name': 'Peyton Turner', 'company': 'Walker Inc'},
    {'name': 'Isaias Fritsch', 'company': 'Walker Inc'},
    {'name': 'Susana Wilderman', 'company': 'Nolan Inc'} ]

print(group_by_company(consultants))
```
